```python
# Step 1: Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Step 2: Load the dataset from GitHub

url = "https://raw.githubusercontent.com/sonukr-62/Digital-Advertising-Campaign-Analysis/main/tech_advertising_campaigns_dataset.csv"

df = pd.read_csv(url)

# Step 3: Display first 5 rows

print("Dataset loaded successfully!")
display(df.head())
```

```python
# Dataset dimensions
print("Total Rows:", df.shape[0])
print("Total Columns:", df.shape[1])

# Column names
print("\nColumn Names:")
print(df.columns.tolist())

# Data types and non-null counts
print("\nDataset Information:")
df.info()

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
print("Platform Distribution:")

platform_counts = df["platform"].value_counts()

display(platform_counts)

# Visualize platform distribution
platform_counts.plot(kind="bar", figsize=(10, 5))

plt.title("Advertising Campaigns by Platform")
plt.xlabel("Platform")
plt.ylabel("Number of Campaigns")
plt.xticks(rotation=45)
plt.show()
```

```python
# Step 4: Dataset Inspection

# 1. Check total rows and columns
print("Dataset Shape:", df.shape)

# 2. Display all column names
print("\nColumn Names:")
print(df.columns.tolist())

# 3. Check data types
print("\nData Types:")
print(df.dtypes)

# 4. Check missing values
print("\nMissing Values:")
display(df.isnull().sum())

# 5. Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())

# 6. Statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

```python
# Check whether required project columns exist

required_columns = [
    "platform",
    "impressions",
    "clicks",
    "conversions",
    "ad_spend",
    "revenue",
    "campaign_name",
    "date",
    "age",
    "gender",
    "city",
    "device"
]

print("PROJECT REQUIREMENTS CHECK\n")

for column in required_columns:
    if column in df.columns:
        print("FOUND:", column)
    else:
        print("MISSING:", column)
```

```python
# Step 6: Calculate Advertising KPIs

# Avoid division by zero
df["CTR"] = np.where(
    df["impressions"] > 0,
    (df["clicks"] / df["impressions"]) * 100,
    0
)

df["CPC"] = np.where(
    df["clicks"] > 0,
    df["ad_spend"] / df["clicks"],
    0
)

df["CPA"] = np.where(
    df["conversions"] > 0,
    df["ad_spend"] / df["conversions"],
    0
)

df["Conversion_Rate"] = np.where(
    df["clicks"] > 0,
    (df["conversions"] / df["clicks"]) * 100,
    0
)

df["ROAS"] = np.where(
    df["ad_spend"] > 0,
    df["revenue"] / df["ad_spend"],
    0
)

# Display calculated metrics
kpi_columns = [
    "platform",
    "impressions",
    "clicks",
    "conversions",
    "ad_spend",
    "revenue",
    "CTR",
    "CPC",
    "CPA",
    "Conversion_Rate",
    "ROAS"
]

display(df[kpi_columns].head(10))
```

```python
# Step 7: Platform-wise Performance Analysis

platform_analysis = df.groupby("platform").agg(
    Total_Impressions=("impressions", "sum"),
    Total_Clicks=("clicks", "sum"),
    Total_Conversions=("conversions", "sum"),
    Total_Spend=("ad_spend", "sum"),
    Total_Revenue=("revenue", "sum")
).reset_index()

# Calculate platform-level KPIs
platform_analysis["CTR"] = (
    platform_analysis["Total_Clicks"] /
    platform_analysis["Total_Impressions"]
) * 100

platform_analysis["CPC"] = (
    platform_analysis["Total_Spend"] /
    platform_analysis["Total_Clicks"].replace(0, np.nan)
)

platform_analysis["CPA"] = (
    platform_analysis["Total_Spend"] /
    platform_analysis["Total_Conversions"].replace(0, np.nan)
)

platform_analysis["Conversion_Rate"] = (
    platform_analysis["Total_Conversions"] /
    platform_analysis["Total_Clicks"].replace(0, np.nan)
) * 100

platform_analysis["ROAS"] = (
    platform_analysis["Total_Revenue"] /
    platform_analysis["Total_Spend"].replace(0, np.nan)
)

# Display results
display(platform_analysis.round(2))
```

```python
# Step 9: Platform Performance Visualizations

# 1. ROAS Comparison
plt.figure(figsize=(10, 5))

plt.bar(
    platform_analysis["platform"],
    platform_analysis["ROAS"]
)

plt.title("Return on Ad Spend (ROAS) by Platform")
plt.xlabel("Advertising Platform")
plt.ylabel("ROAS")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()


# 2. Cost Per Acquisition Comparison
plt.figure(figsize=(10, 5))

plt.bar(
    platform_analysis["platform"],
    platform_analysis["CPA"]
)

plt.title("Cost Per Acquisition (CPA) by Platform")
plt.xlabel("Advertising Platform")
plt.ylabel("CPA (Rs.)")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()


# 3. Total Revenue vs Total Advertising Spend
x = np.arange(len(platform_analysis))
width = 0.35

plt.figure(figsize=(12, 5))

plt.bar(
    x - width/2,
    platform_analysis["Total_Spend"],
    width,
    label="Advertising Spend"
)

plt.bar(
    x + width/2,
    platform_analysis["Total_Revenue"],
    width,
    label="Revenue"
)

plt.title("Advertising Spend vs Revenue by Platform")
plt.xlabel("Advertising Platform")
plt.ylabel("Amount (Rs.)")
plt.xticks(x, platform_analysis["platform"], rotation=45)
plt.legend()
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# Step 11: Marketing Budget Recommendation Engine

total_budget = 1000000

# Calculate total ROAS
total_roas = platform_analysis["ROAS"].sum()

# Calculate budget allocation percentage
platform_analysis["Budget_Percentage"] = (
    platform_analysis["ROAS"] / total_roas
) * 100

# Calculate recommended budget
platform_analysis["Recommended_Budget"] = (
    platform_analysis["Budget_Percentage"] / 100
) * total_budget

# Round values
platform_analysis["Budget_Percentage"] = (
    platform_analysis["Budget_Percentage"].round(2)
)

platform_analysis["Recommended_Budget"] = (
    platform_analysis["Recommended_Budget"].round(2)
)

# Display recommendations
display(
    platform_analysis[
        ["platform", "ROAS", "Budget_Percentage", "Recommended_Budget"]
    ]
)
```

```python
# Budget Allocation Visualization

plt.figure(figsize=(10, 6))

plt.pie(
    platform_analysis["Recommended_Budget"],
    labels=platform_analysis["platform"],
    autopct="%1.1f%%",
    startangle=140
)

plt.title("Recommended Advertising Budget Allocation (Rs. 1,000,000)")
plt.show()
```

```python
# Verify total budget allocation

print("Total Recommended Budget:")
print(platform_analysis["Recommended_Budget"].sum())

print("\nBudget Distribution Check:")
print(
    "Total Percentage:",
    platform_analysis["Budget_Percentage"].sum(),
    "%"
)
```

```python
# Step 12: Enhanced Budget Recommendation Engine

# Normalize ROAS (higher is better)
platform_analysis["ROAS_Score"] = (
    platform_analysis["ROAS"] /
    platform_analysis["ROAS"].max()
)

# Normalize Conversion Rate (higher is better)
platform_analysis["Conversion_Score"] = (
    platform_analysis["Conversion_Rate"] /
    platform_analysis["Conversion_Rate"].max()
)

# Normalize CPA (lower is better)
platform_analysis["CPA_Score"] = (
    platform_analysis["CPA"].min() /
    platform_analysis["CPA"]
)

# Calculate combined performance score
platform_analysis["Performance_Score"] = (
    platform_analysis["ROAS_Score"] * 0.50 +
    platform_analysis["Conversion_Score"] * 0.30 +
    platform_analysis["CPA_Score"] * 0.20
)

# Calculate enhanced budget allocation
total_budget = 1000000

platform_analysis["Enhanced_Budget_Percentage"] = (
    platform_analysis["Performance_Score"] /
    platform_analysis["Performance_Score"].sum()
) * 100

platform_analysis["Enhanced_Recommended_Budget"] = (
    platform_analysis["Enhanced_Budget_Percentage"] /
    100
) * total_budget

# Display results
display(
    platform_analysis[
        [
            "platform",
            "ROAS_Score",
            "Conversion_Score",
            "CPA_Score",
            "Performance_Score",
            "Enhanced_Budget_Percentage",
            "Enhanced_Recommended_Budget"
        ]
    ].round(2)
)
```

```python
# Compare Original and Enhanced Budget Allocation

budget_comparison = platform_analysis[
    [
        "platform",
        "Recommended_Budget",
        "Enhanced_Recommended_Budget"
    ]
].copy()

budget_comparison = budget_comparison.rename(columns={
    "Recommended_Budget": "ROAS_Only_Budget",
    "Enhanced_Recommended_Budget": "Enhanced_Budget"
})

display(budget_comparison.round(2))
```

```python
# Step 15.2: Pandas Benchmark

import pandas as pd
import time
import os

# Prepare CSV files for smaller dataset sizes
benchmark_folder = "/content/pandas_benchmark"
os.makedirs(benchmark_folder, exist_ok=True)

original_df.to_csv(
    benchmark_folder + "/data_10k.csv", index=False
)

df_100k.to_csv(
    benchmark_folder + "/data_100k.csv", index=False
)

df_1m.to_csv(
    benchmark_folder + "/data_1m.csv", index=False
)

# Paths of all four datasets
dataset_paths = {
    "10K": benchmark_folder + "/data_10k.csv",
    "100K": benchmark_folder + "/data_100k.csv",
    "1M": benchmark_folder + "/data_1m.csv",
    "10M": "/content/drive/MyDrive/BigData_Project/advertising_10M.csv"
}

# Columns to aggregate
metrics = [
    "impressions",
    "clicks",
    "conversions",
    "ad_spend",
    "revenue"
]

# Pandas chunk-processing function
def pandas_benchmark(file_path):

    start_time = time.perf_counter()

    final_result = None
    max_chunk_memory = 0

    for chunk in pd.read_csv(file_path, chunksize=100_000):

        # Measure memory of current chunk
        chunk_memory = (
            chunk.memory_usage(deep=True).sum() / (1024 ** 2)
        )

        max_chunk_memory = max(max_chunk_memory, chunk_memory)

        # Aggregate campaign metrics by platform
        partial_result = chunk.groupby("platform")[metrics].sum()

        if final_result is None:
            final_result = partial_result
        else:
            final_result = final_result.add(
                partial_result, fill_value=0
            )

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return final_result, execution_time, max_chunk_memory


# Run benchmark on all dataset sizes
benchmark_results = []

for size, path in dataset_paths.items():

    print(f"\nProcessing {size} dataset...")

    result, execution_time, chunk_memory = pandas_benchmark(path)

    benchmark_results.append({
        "Dataset_Size": size,
        "Execution_Time_Seconds": round(execution_time, 2),
        "Maximum_Chunk_Memory_MB": round(chunk_memory, 2)
    })

    print("Execution Time:", round(execution_time, 2), "seconds")
    print("Maximum Chunk Memory:", round(chunk_memory, 2), "MB")

# Display final benchmark table
pandas_results = pd.DataFrame(benchmark_results)

print("\nPANDAS PERFORMANCE RESULTS")
display(pandas_results)
```

```python
# Step 15.3: MapReduce-style Benchmark

import pandas as pd
import time

metrics = [
    "impressions",
    "clicks",
    "conversions",
    "ad_spend",
    "revenue"
]

# MAPPER: Process one chunk and produce platform-wise totals
def mapper(chunk):

    mapped_data = chunk.groupby("platform")[metrics].sum()

    return mapped_data


# REDUCER: Combine results produced by all mappers
def reducer(final_result, mapped_result):

    if final_result is None:
        return mapped_result

    return final_result.add(mapped_result, fill_value=0)


# MapReduce processing function
def mapreduce_benchmark(file_path):

    start_time = time.perf_counter()

    final_result = None
    max_chunk_memory = 0

    # Read dataset in chunks
    for chunk in pd.read_csv(file_path, chunksize=100_000):

        chunk_memory = (
            chunk.memory_usage(deep=True).sum() / (1024 ** 2)
        )

        max_chunk_memory = max(
            max_chunk_memory, chunk_memory
        )

        # Map stage
        mapped_result = mapper(chunk)

        # Reduce stage
        final_result = reducer(
            final_result, mapped_result
        )

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return final_result, execution_time, max_chunk_memory


# Run benchmark on all dataset sizes
mapreduce_benchmark_results = []

for size, path in dataset_paths.items():

    print(f"\nProcessing {size} dataset using MapReduce...")

    result, execution_time, chunk_memory = (
        mapreduce_benchmark(path)
    )

    mapreduce_benchmark_results.append({
        "Dataset_Size": size,
        "Execution_Time_Seconds": round(execution_time, 2),
        "Maximum_Chunk_Memory_MB": round(chunk_memory, 2)
    })

    print("Execution Time:", round(execution_time, 2), "seconds")
    print("Maximum Chunk Memory:", round(chunk_memory, 2), "MB")


# Display results
mapreduce_results = pd.DataFrame(
    mapreduce_benchmark_results
)

print("\nMAPREDUCE PERFORMANCE RESULTS")

display(mapreduce_results)
```

```python
import time
import pandas as pd

# Dataset paths
dataset_paths = {
    "10K": "/content/pandas_benchmark/data_10k.csv",
    "100K": "/content/pandas_benchmark/data_100k.csv",
    "1M": "/content/pandas_benchmark/data_1m.csv",
    "10M": "/content/drive/MyDrive/BigData_Project/advertising_10M.csv"
}

metrics = [
    "impressions",
    "clicks",
    "conversions",
    "ad_spend",
    "revenue"
]

spark_results = []

for size, path in dataset_paths.items():

    print(f"\nProcessing {size} dataset using Spark...")

    start_time = time.perf_counter()

    # Read CSV using Spark
    spark_df = spark.read.csv(
        path,
        header=True,
        inferSchema=True
    )

    # Aggregate metrics by platform
    result = spark_df.groupBy("platform").agg(
        *[
            __import__("pyspark").sql.functions.sum(col).alias(col)
            for col in metrics
        ]
    )

    # Execute aggregation
    result.collect()

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    # Estimate input data size in MB
    import os
    file_size_mb = os.path.getsize(path) / (1024 ** 2)

    spark_results.append({
        "Dataset_Size": size,
        "Execution_Time_Seconds": round(execution_time, 2),
        "Input_File_Size_MB": round(file_size_mb, 2)
    })

    print("Execution Time:", round(execution_time, 2), "seconds")
    print("Input File Size:", round(file_size_mb, 2), "MB")

# Display results
spark_results_df = pd.DataFrame(spark_results)

print("\nSPARK PERFORMANCE RESULTS")
display(spark_results_df)
```
